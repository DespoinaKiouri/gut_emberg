```python
import pandas as pd
```

```python
train_path = "./data/train.csv"
val_path = "./data/val.csv"
test_path = "./data/test.csv"
```

```python
# Read the CSV files
train_df = pd.read_csv(train_path)
val_df = pd.read_csv(val_path)
test_df = pd.read_csv(test_path)

# Print the first few rows of each DataFrame
print("Train DataFrame:")
print(train_df.head())
print("\nValidation DataFrame:")
print(val_df.head())
print("\nTest DataFrame:")
print(test_df.head())
```

```python

proteins = list()

for df in [train_df, val_df, test_df]:
    for p1 in df["P1"]:
        proteins.append(p1)
    for p2 in df["P2"]:
        proteins.append(p2)

proteins = set(proteins)

print(len(proteins))
proteins = list(proteins)
```

```python
# Slice the proteins to batches of 90000 and save to txt
# We need this to fetch uniprot information (e.g. genes, sequences)
for i in range(0, len(proteins), 90000):
    with open(f"./data/proteins_{i}.txt", "w") as f:
        f.write("\n".join(proteins[i : i + 90000]))
```
