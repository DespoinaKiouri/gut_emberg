```python
import os
import pandas as pd

from c3ppi.ppi_predictor.trainer.train import Trainer
from c3ppi.ppi_predictor.evaluator.evaluate import Evaluator
```

```python
emb_dir = "./data/embeddings"

train_path = "./data/train.csv"
val_path = "./data/val.csv"
test_path = "./data/test.csv"
```

```python
# Read and filter if embeddings exist
train_df = pd.read_csv(train_path)
val_df = pd.read_csv(val_path)
test_df = pd.read_csv(test_path)

proteins_with_embeddings = [p.split("_")[0] for p in os.listdir(emb_dir)]

train_df = train_df[train_df["P1"].isin(proteins_with_embeddings) & train_df["P2"].isin(proteins_with_embeddings)]
val_df = val_df[val_df["P1"].isin(proteins_with_embeddings) & val_df["P2"].isin(proteins_with_embeddings)]
test_df = test_df[test_df["P1"].isin(proteins_with_embeddings) & test_df["P2"].isin(proteins_with_embeddings)]

train_df.to_csv("./data/filtered_ppi_data_train.csv", index=False)
val_df.to_csv("./data/filtered_ppi_data_val.csv", index=False)
test_df.to_csv("./data/filtered_ppi_data_test.csv", index=False)
```

```python
trainer_config = {
    "train_data_path": "./data/filtered_ppi_data_train.csv",
    "val_data_path": "./data/filtered_ppi_data_val.csv",
    "embeddings_dir": "./data/embeddings",
    "save_name": "c3ppi_model",
    "batch_size": 512
}

trainer = Trainer(**trainer_config)
trainer.run()
```

```python
val_config = {
    "mode": "val",
    "data_path": "./data/filtered_ppi_data_val.csv",
    "embeddings_dir": "./data/embeddings",
    "model_name": "c3ppi_model",
    "batch_size": 512
}

evaluator = Evaluator(**val_config)
evaluator.run()
```

```python
val_config = {
    "mode": "test",
    "data_path": "./data/filtered_ppi_data_val.csv",
    "embeddings_dir": "./data/embeddings",
    "model_name": "c3ppi_model",
    "threshold": 0.4261,
    "batch_size": 512
}

evaluator = Evaluator(**val_config)
evaluator.run()
```
