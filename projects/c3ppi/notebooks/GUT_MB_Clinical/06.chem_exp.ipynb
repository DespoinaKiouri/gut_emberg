```python
import pandas as pd
```

```python
file1 = "/home/c3biolab/c3biolab_projects/doctorals/DPK/c3ppi/notebooks/GUT_MB_Clinical/data/idmapping_chem_exp.tsv"
file2 = "/home/c3biolab/c3biolab_projects/doctorals/DPK/c3ppi/notebooks/GUT_MB_Clinical/data/idmapping_chem_exp_2.tsv"
file_nfkb = "/home/c3biolab/c3biolab_projects/doctorals/DPK/c3ppi/notebooks/GUT_MB_Clinical/data/idmapping_chem_exp_NFKB.tsv"
```

```python
df1 = pd.read_csv(file1, sep="\t", header=0)
df1
```

```python
file2 = pd.read_csv(file2, sep="\t", header=0)
file2
```

```python
df3 = pd.read_csv(file_nfkb, sep="\t", header=0)
df3
```

```python
# Select rows that contain "NFKB" in the "Gene Names" column
df_nfkb = df3[df3["Gene Names"].str.contains("NFKB", case=False, na=False)]
df_nfkb["From"] = "NFKB"
```

```python
# Concatenate the three dataframes - keep only From Entry	Protein names	Gene Ontology (molecular function)	Gene Ontology (biological process)
df_combined = pd.concat([df1[["From", "Entry", "Protein names", "Gene Ontology (molecular function)", "Gene Ontology (biological process)"]],
                         file2[["From", "Entry", "Protein names", "Gene Ontology (molecular function)", "Gene Ontology (biological process)"]],
                         df_nfkb[["From", "Entry", "Protein names", "Gene Ontology (molecular function)", "Gene Ontology (biological process)"]]])
df_combined
```

```python
hum_degree_file = "./data/human_protein_degrees.csv"
hum_degree = pd.read_csv(hum_degree_file)
hum_degree = hum_degree.set_index("protein")
hum_degree
```

```python
# Merge the chemical experiment data with the human protein degree data
merged_df = df_combined.merge(hum_degree, left_on="Entry", right_index=True, how="left")
merged_df = merged_df.dropna(subset=["degree_human"])

# Rename columns for clarity
merged_df = merged_df.rename(columns={"From": "Gene", "Entry": "Protein", "degree_human": "Degree"})
merged_df = merged_df.reset_index(drop=True)
merged_df = merged_df.reset_index(drop=True)
merged_df
```

```python
# For each human protein in the merged dataframe, read the parquet file with its interactions under notebooks/GUT_MB_Clinical/data/inference_results/per_protein_results and obtain a random bacterial interactor as an example.
# Add it as extra columns to the merged dataframe and save it as a new dataframe.
interactor_list = []
for protein in merged_df["Protein"]:
    parquet_file = f"./data/inference_results/per_protein_results/{protein}_predictions.parquet"
    interactions_df = pd.read_parquet(parquet_file)
    if not interactions_df.empty:
        random_interactor = interactions_df.sample(n=1)
        interactor_list.append(random_interactor["bacterial_protein"].values[0])
    else:
        interactor_list.append(None)
merged_df["Random_Bacterial_Interactor"] = interactor_list
merged_df.to_csv("./data/merged_chem_exp_human_degree.csv", index=False)
```
