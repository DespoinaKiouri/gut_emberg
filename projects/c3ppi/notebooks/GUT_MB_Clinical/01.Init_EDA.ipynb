## Find abundances of species from OTU table
*OTU table file organization*
* First column: Taxonomy ID
* Last column: Taxonomy description (in this study, k__:kingdom, p__:phylum, c__:class, o__:order, f__:family, g__:genus)
* All the other columns: Relative abundances
* First row: Sample IDs

Final Result: 179 genus

```python
# Read student file
student_df = pd.read_csv(student, sep="\t")
# Unnamed: 0 rename
student_df = student_df.rename(columns={'Unnamed: 0':"IDs"})
# Keep all the lines with 6 ";" in taxonomy column
student_df = student_df[student_df["taxonomy"].str.count(";") == 6]
# Keep only unique 
student_df = student_df.drop_duplicates(subset=["taxonomy"])
# Split taxonomy after fifth ";" and keep second part
student_df["Genus"] = student_df["taxonomy"].str.split(";").str[5]
# Drop taxonomy column
student_df = student_df.drop(columns=["taxonomy"])
# Drop  cells that contain " " in Genus columns
student_df = student_df[~student_df["Genus"].str.contains(" ")]

student_df
```

```python
student_df.to_csv("./data/OTUs_Table_students_FILTERED.csv")
# Create list of Student Sample IDs
student_samples = student_df.columns
# Delete first column
student_samples = student_samples[1:]
# Delete last column
student_samples = student_samples[:-1]
student_samples= student_samples.to_list()
```

```python
# Read patients file
patients_df = pd.read_csv(patients, sep="\t")
# Unnamed: 0 rename
patients_df = patients_df.rename(columns={'Unnamed: 0':"IDs"})
# Keep all the lines with 6 ";" in taxonomy column
patients_df = patients_df[patients_df["taxonomy"].str.count(";") == 6]
# Keep only unique 
patients_df = patients_df.drop_duplicates(subset=["taxonomy"])
# Split taxonomy after fifth ";" and keep second part
patients_df["Genus"] = patients_df["taxonomy"].str.split(";").str[5]
# Drop taxonomy column
patients_df = patients_df.drop(columns=["taxonomy"])
# Drop  cells that contain " " in Genus columns
patients_df = patients_df[~patients_df["Genus"].str.contains(" ")]

patients_df
```

```python
# To csv
patients_df.to_csv("./data/OTUs_Table_patients_FILTERED.csv")
# Create list of Patiebts Sample IDs
patients_samples = patients_df.columns
# Delete first column
patients_samples = patients_samples[1:]
# Delete last column
patients_samples = patients_samples[:-1]
patients_samples= patients_samples.to_list()
```

## Find abundances of species from taxonomy file
* Take everything that starts from g__ as a genus

Final result: 204 genus

```python
# Read tax file
tax_df = pd.read_csv(tax, sep="\t")
tax_df = tax_df.rename(columns={'Unnamed: 0': 'Taxonomy'})
# Keep only rows with Taxonomy that start with g__
tax_df = tax_df[tax_df["Taxonomy"].str.startswith("g__")]
# Filter columns of tax_df by student_samples
students_samples_abundance = tax_df[["Taxonomy"]+student_samples] 
# Filter tax_df by patients_samples
patients_samples_abundance = tax_df[["Taxonomy"]+patients_samples]
```

```python
students_samples_abundance = students_samples_abundance.reset_index(drop=True)
students_samples_abundance["Average"] = students_samples_abundance[student_samples].mean(axis=1)
students_samples_abundance["Average"].describe()
```

```python
patients_samples_abundance = patients_samples_abundance.reset_index(drop=True)
patients_samples_abundance["Average"] = patients_samples_abundance[patients_samples].mean(axis=1)
patients_samples_abundance["Average"].describe()
```

```python
import pandas as pd
import numpy as np
from scipy.stats import mannwhitneyu
from statsmodels.stats.multitest import multipletests

# Remove 'Average' column if present
students_samples_abundance = students_samples_abundance.drop(columns=['Average'], errors='ignore')
patients_samples_abundance = patients_samples_abundance.drop(columns=['Average'], errors='ignore')

# Reshape dataframes to long format
students_long = students_samples_abundance.melt(
    id_vars='Taxonomy',
    var_name='SampleID',
    value_name='Abundance'
)
students_long['Phenotype'] = 'Healthy'

patients_long = patients_samples_abundance.melt(
    id_vars='Taxonomy',
    var_name='SampleID',
    value_name='Abundance'
)
patients_long['Phenotype'] = 'Cancer'

# Concatenate
combined = pd.concat([students_long, patients_long], ignore_index=True)

# Handle duplicates: prioritize Cancer if sample in both
combined = combined.sort_values(by='Phenotype', ascending=True)
combined = combined.drop_duplicates(subset=['Taxonomy', 'SampleID'], keep='last')

# Prepare output
results = []

# Per Taxonomy
for taxon, group in combined.groupby('Taxonomy'):
    healthy_vals = group.loc[group['Phenotype'] == 'Healthy', 'Abundance']
    cancer_vals = group.loc[group['Phenotype'] == 'Cancer', 'Abundance']
    
    if (healthy_vals.sum() == 0) and (cancer_vals.sum() == 0):
        continue
    
    try:
        stat, p_value = mannwhitneyu(
            healthy_vals,
            cancer_vals,
            alternative='two-sided'
        )
    except ValueError:
        p_value = 1.0

    median_healthy = healthy_vals.median()
    median_cancer = cancer_vals.median()

    results.append({
        'Taxonomy': taxon,
        'Median_Healthy': median_healthy,
        'Median_Cancer': median_cancer,
        'p_value': p_value
    })

# Convert to DataFrame
results_df = pd.DataFrame(results)

# Multiple testing correction
reject, qvals, _, _ = multipletests(results_df['p_value'], method='fdr_bh')
results_df['q_value'] = qvals
results_df['Significant'] = reject

# Assign phenotype label
def assign_label(row):
    if row['q_value'] < 0.05:
        if row['Median_Healthy'] > row['Median_Cancer']:
            return 'Healthy-associated'
        elif row['Median_Cancer'] > row['Median_Healthy']:
            return 'CRC-associated'
        else:
            return 'Both (No clear association)'
    else:
        return 'Both (No clear association)'

results_df['Phenotype_Assignment'] = results_df.apply(assign_label, axis=1)

# Sort
results_df = results_df.sort_values(by='q_value')

# Save or display
results_df.to_csv("./data/bacterial_phenotype_assignment.csv", index=False)

# Show top rows
results_df.reset_index(drop=True)
```

```python
# CRC-associated
crc_associated = results_df[results_df['Phenotype_Assignment'] == 'CRC-associated']
crc_associated = crc_associated[['Taxonomy', 'Median_Cancer', 'p_value', 'q_value']].reset_index(drop=True)
crc_associated["Median_Cancer"] = crc_associated["Median_Cancer"].map(lambda x: f"{x:.4f}")
crc_associated["p_value"] = crc_associated["p_value"].map(lambda x: f"{x:.3e}")
crc_associated["q_value"] = crc_associated["q_value"].map(lambda x: f"{x:.3e}")
# Save CRC
crc_associated.to_excel("./data/bacterial_crc_associated.xlsx", index=False)
```

```python
# Healthy-associated
healthy_associated = results_df[results_df['Phenotype_Assignment'] == 'Healthy-associated']
healthy_associated = healthy_associated[['Taxonomy', 'Median_Healthy', 'p_value', 'q_value']].reset_index(drop=True)
# Keep 3 decimal places
healthy_associated["Median_Healthy"] = healthy_associated["Median_Healthy"].map(lambda x: f"{x:.4f}")
healthy_associated["p_value"] = healthy_associated["p_value"].map(lambda x: f"{x:.3e}")
healthy_associated["q_value"] = healthy_associated["q_value"].map(lambda x: f"{x:.3e}")
# Save Healthy
healthy_associated.to_excel("./data/bacterial_healthy_associated.xlsx", index=False)
```
